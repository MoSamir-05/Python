### Write a python program to learn use of python lambda functions.

```python
add = lambda a, b: a + b
print("Addition:", add(10, 5))
square = lambda x: x * x
print("Square:", square(6))
numbers = [1, 2, 3, 4, 5, 6]
even_numbers = list(filter(lambda n: n % 2 == 0, numbers))
print("Even numbers:", even_numbers)
double = list(map(lambda n: n * 2, numbers))
print("Double list:", double)
words = ["apple", "mango", "banana"]
sorted_words = sorted(words, key=lambda w: w[-1])
print("Sorted by last letter:", sorted_words)
```
