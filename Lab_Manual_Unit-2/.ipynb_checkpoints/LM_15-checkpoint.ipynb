### 15.1: Magicians: Make a list of magician's names. Pass the list to a function called show_magicians(), which prints the name of each magician in the list.

```python
def show_magicians(magicians):
    for name in magicians:
        print(name)
magician_list = ["Harry Houdini", "David Blaine", "Criss Angel", "Dynamo"]
show_magicians(magician_list)
```

### 15.2: Great Magicians: Start with a copy of your program from Exercise 8-9. Write a function called make_great() that modifies the list of magicians by add- ing the phrase the Great to each magician's name. Call show_magicians() to see that the list has actually been modified.

```python
def show_magicians(magicians):
    for name in magicians:
        print(name)

def make_great(magicians):
    
    for i in range(len(magicians)):
        magicians[i] = magicians[i] + " the Great"


magician_list = ["Harry Houdini", "David Blaine", "Criss Angel", "Dynamo"]


make_great(magician_list)


show_magicians(magician_list)
```
