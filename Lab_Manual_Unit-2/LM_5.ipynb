### Write a Python Program to learn use of Python Dictionary methods

```python
student = {
    "name": "Samir",
    "age": 20,
    "course": "MCA"
}

print("Original Dictionary:")
print(student)

print("\nKeys in dictionary:")
print(student.keys())

print("\nValues in dictionary:")
print(student.values())

print("\nItems (key-value pairs):")
print(student.items())

print("\nGet value of 'name':")
print(student.get("name"))

student.update({"age": 21, "city": "Anand"})
print("\nAfter update() method:")
print(student)

student.pop("course")
print("\nAfter pop('course'):")
print(student)

student.popitem()
print("\nAfter popitem() (removes last item):")
print(student)

student.clear()
print("\nAfter clear():")
print(student)
```
