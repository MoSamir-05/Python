## Slicing

```python
tp1=(1,2,3,4,5)
print(tp1[2:4])
print(tp1[::-1])
```

## Concatenation

```python
tp1=(1,2,3,4)
tp2=("sam","kaifu","hujju","saku")
# print(tp1+tp2)
for a ,b in zip(tp1,tp2):
    print(a,b ,end="\n")
```
