## List Creation


```python
lst = ["apple","banana","cherry"]
nums = [1,2,3]
print(nums)
print(lst)
```

### Length of list

```python
print(len(lst))
print(type(lst))
```

# Access list items

```python
print(lst[2])
print(lst[-1])
print(lst[-3:-1])
print(lst[::-1])
```

## + and * operator

```python
lst100 = [1,2,3]
lst101 = [3,4,5]
print(lst100 + lst100)
print(lst100 * 3)
```
