
# String Methods

### 1.Capitalize()

```python
str = "hello world"
print(str.capitalize())
```

### 2. lower()

```python
str1 = "HELLO WORLD"
print(str1.lower())
```

### 3. title()

```python
str2 = "hello world"
print(str2.title())
```

### 4.isalnum()

```python
str3 = "hello world@0505"
print(str3.isalnum())

str4 = "helloworld12"
print(str4.isalnum())
print(str3[1])
print(str3[-1])
print(str3[1:4])
```

### 5. isalpha()

```python
str5 = "hello"
str6 = "1234"
print(str5.isalpha())
print(str6.isalpha())
```

### 6. isdigit()

```python
str7 = 123
str8 = "python"
print(str7.isdigit())
print(str8.isdigit())
```

### 7. islower()

```python
str9 = "hello Python"
str10 = "kaif "
print(str10.islower())
print(str9.islower())
```

### 8. isupper()

```python
str11 = "PYTHON"
str12 = "python"
print(str11.isupper())
print(str12.isupper())
```

### 9. upper()

```python
str13 = "rehan"
print(str13.upper())
```

### 10.swapcase()

```python
str14 = "mOOn"
print(str14.swapcase())
```

### 11.join()

```python
str15 = "hello"
str16 = "world"
sep = ","
str17 = sep.join(str15)
print(str17)
```

### 12. Len()

```python
str17 = "Bye"
print(len(str17))
```

### 13. strip()

```python
str18 = "###hey###"
print(str18.strip("#"))
```
