## Operators

### 1.Arithmetic Operators

```python
a, b = map(int,input("Enter 2 numbers :").split())

print("Addition : ",a+b)
print("Substraction : ", a - b)
print("Multiplication :", a * b)
print("Division : ", a / b)
print("Modulous :", a % b)
```

### 2.Logical Operators

```python
a,b = map(int,input("enter 2 numbers :").split())
if a > b :
    print(f"{a} is gt {b}")
elif a < b:
    print(f"{a} is lt {b}")
print(not(a is b))
```

### 3.Assignment Operators

```python
a = 5
b = True
c = 4.5
a += a
print(a)
print(b)
print(c)
```
