# Tuple

```python
tp = (1,2,3,4,5)
tp1 = ("hello","world")
print(tp)
print(tp1)
print(type(tp))
tp2 = (0.9,0.4,0.20,0.1)
print(tp2)
single_tuple = (1,)
print("single tuple",single_tuple)
```

## Tuple concatenation

```python
tp91 = (1,2)
tp92 = (3,4)
print(tp91 + tp92)
print(tp91 * 2)
```

## Changing tuple elems

```python
tp[0] = 100
print(tp)
```

## Removing tuple elems

```python
tuple50 = (100,200,300,400,500)
lst = list(tuple50)
lst.remove(400)
tuple51 = tuple(lst)
print(tuple51)
```

## delete entire tuple 

```python
tuple52 = (1,2,3)
del tuple52
print(tuple52)
```

# Tuple Slicing

```python
tp = (1,2,3,4,5,6,7,8,9)
print("accesing first elem",tp[1])
print("accessing last elem",tp[-1])
print("accessing from 2nd index to 5th",tp[2:6])
print("reverse",tp[::-1])
print("negative indexing ",tp[-1:-3])
print("indexing with jump",tp[1:8:2])
print("omit start",tp[:5])
print("omit end",tp[2:])
print("negative with omit end",tp[-3:])
print("negative ibndex with omit start",tp[:-2])
print("jumping tuple without end",tp[1::2])
print("jumping tuple without start-end",tp[::2])
```
