## printing Dictionary

```python
print("Mo Samir")
dict1={1:"sam",2:"kaifu",3:"saku",4:"hujju",5:"Dombhai"}
print(dict1)
print(dict1[4])
```

## accessing Dictionary

```python
print("Mo Samir")
print(dict1[3])
key1=dict1[4]
print(key1)
```

#### useing get method

```python
print("Mo Samir")
print(dict1.get(3))
```

#### value method

```python
print("Mo Samir")
print(dict1.values())
```

#### print only keys

```python
print("Mo Samir")
print(dict1.keys())
```

#### print only items

```python
print("Mo Samir")
print(dict1.items())
```

## change element

```python
print("Mo Samir")
dict1[1]="babu bhai"
print(dict1)
```

#### update method

```python
print("Mo Samir")
dict1.update({1:"sam"})
print(dict1)
```

#### add new element

```python
print("Mo Samir")
dict1[6]="salman"
print(dict1)
```

#### Remove element

```python
print("Mo Samir")
dict1.pop(1)
print(dict1)
```

```python
print("Mo Samir")
dict1.popitem()
print(dict1)
print(dict1.clear())
```
