```python
birthday={
    "samir":"05-03-2004",
    "kaif":"18-08-2003",
    "hujaifa":"12-9-2005"
}
name=input("enter your name to check your birthdate : ").lower();
if name in birthday:
    bday=birthday[name]
    parts=bday.split("-")
    print("date",parts[0])
    print("Month",parts[1])
    print("year",parts[2])
    join_date="/".join(parts)
    print("formate birthday : ",join_date)
else:
    print("No birthday found :(")
    
    
```
