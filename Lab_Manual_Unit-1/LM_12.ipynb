```python
per = float(input("Enter your percentage : "))
att = float(input("Enter Your Attendance : "))

if per >= 90:
    if att >= 80:
        print("Your Grade is A+")
    else:
        print("Your Grade is A-")
elif per >= 80:
    if att >= 80:
        print("Your Grade is B+")
    else:
        print("Your Grade is B-")
elif per >= 70:
    if att >= 80:
        print("Your Grade is C+")
    else:
        print("Your Grade is C-")
else:
    print("Your Grade is F")
```
