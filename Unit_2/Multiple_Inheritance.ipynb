```python
class Person:
    def __init__(self,name,age):
        self.name=name
        self.age=age
    def show_Person(self):
        return print("Name  : ",self.name,
                     "\nAge : ",self.age )

class job:
    def __init__(self,title,salary):
        self.title=title
        self.salary=salary
    def show_job (self):
        return print("title    : ",self.title,
                     "\nsalary : ",self.salary)

class manager(Person,job):
    def __init__(self,name,age,title,salary,department):
        Person.__init__(self,name,age)
        job.__init__(self,title,salary)
        self.department=department
    def show_manager(self):
        self.show_Person()
        self.show_job()
        return print("department : ",self.department)

c=manager("samir",20,"pro manager",100000,"it")
c.show_manager()
```
